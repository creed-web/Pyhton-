```python
#Polymorphism is when a object from the parent class can be performed diffrently in it's child classes.

#Parent class
class animal:
    def speak(self):
        return ("The animal is speaking...")
    
#derived class 1
class dog(animal):
    def speak(self):
        return "woof!"
    
#derived class 2
class cat(animal):
    def speak(self):
        return "meow!"
    
#creating objects    
dog = dog()
print(dog.speak())
cat = cat()
print(cat.speak())

#function that demonstrates polymorphism 
def animal_speak(animal):
    print(animal.speak())

animal_speak(cat)
```

```python
#Polymorphism with function and methods

#base class
class Shape:
    def area(self):
        return "area of the shape is: "
    
class Rectangle(Shape):
    def __init__(self,height,width):
        self.height = height
        self.width = width
        
    def area(self):
        return (f"Area of the rectangle is {self.height*self.width}")
    
class circle(Shape):
    def __init__(self,radius):
        self.radius = radius

    def area(self):
        return (f"Area of the circle is : {3.14 * self.radius * self.radius}")

#creating objects    
Rect = Rectangle(4,4)
Cir= circle(4)
print(Rect.area())
print(Cir.area())
```
